```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = np.loadtxt('diabetes.csv', delimiter=',', skiprows=1, usecols=(1, 5, 8))
X = data[:, :2]   # Glucose, BMI
y = data[:, 2]    # Outcome (0 veya 1)
m = X.shape[0]

print(f"X SHAPE: {X.shape}")
print(f"Y SHAPE: {y.shape}")
```

```python
def plotData(X, y):
    plt.figure(figsize=(8, 6))
    pos = (y == 1)
    neg = (y == 0)
    
    plt.scatter(X[pos, 0], X[pos, 1], c='b', marker='+', label="Outcome = 1 (Diyabet)")
    plt.scatter(X[neg, 0], X[neg, 1], c='r', marker='o', label="Outcome = 0 (Sağlıklı)", alpha=0.6)
    
    plt.xlabel('Glucose')
    plt.ylabel('BMI')
    plt.title('Diabetes Dataset (2 Özellik)')
    plt.legend()
    plt.grid(True)
    plt.show()

plotData(X, y)
```

```python
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)
X_std[X_std == 0] = 1.0 
X = (X - X_mean) / X_std

# Bias sütunu ekle
Xb = np.hstack((np.ones((X.shape[0], 1)), X))

print("Normalized Xb shape:", Xb.shape)
print("Y shape:", y.shape)
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def costFunction(theta, X, y, lam):
    m = len(y)
    h = sigmoid(np.dot(X, theta))
    J = -(np.dot(y, np.log(h)) + np.dot(1 - y, np.log(1 - h))) / m
    J += lam * np.sum(theta[1:] ** 2) / (2 * m)
    grad = np.dot(X.T, h - y) / m
    grad[1:] += lam * theta[1:] / m
    return J, grad


def gradientDescent(X, y, theta, alpha, num_iters, lam=0):
    np.seterr(divide='ignore', invalid='ignore')
    J_history = []
    for i in range(num_iters):
        J, grad = costFunction(theta, X, y, lam)
        theta -= alpha * grad
        J_history.append(J)
    return theta, J_history
```

```python
theta0 = np.zeros(X.shape[1])
alpha = 0.01
iteration_values = [200, 800, 1000]
cost_histories = {}

for num_iters in iteration_values:
    theta_trained, J_history = gradientDescent(X, y, theta.copy(), alpha, num_iters)
    cost_histories[num_iters] = J_history
    print(f"Final cost for {num_iters} iterations: {J_history[-1]}")

best_iter = min(cost_histories, key=lambda k: cost_histories[k][-1])
print(f"\nEn iyi iterasyon: {best_iter} (final cost = {cost_histories[best_iter][-1]:.6f})")
```

```python
theta0 = np.zeros(Xb.shape[1]) 

plt.figure(figsize=(10, 6))

for num_iters in iteration_values:
    theta_trained, _ = gradientDescent(Xb, y, theta0.copy(), alpha, num_iters)

    x_values = np.array([Xb[:, 1].min(), Xb[:, 1].max()])
    if not np.isclose(theta_trained[2], 0.0):
        y_values = -(theta_trained[0] + theta_trained[1] * x_values) / theta_trained[2]
        plt.plot(x_values, y_values, label=f'{num_iters} iterations')

plt.scatter(Xb[y == 1][:, 1], Xb[y == 1][:, 2], label="Outcome = 1", marker='o', color='b')
plt.scatter(Xb[y == 0][:, 1], Xb[y == 0][:, 2], label="Outcome = 0", marker='x', color='r')
plt.xlabel("Normalized Glucose")
plt.ylabel("Normalized BMI")
plt.title("Decision Boundaries for Different Iterations")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
best_num_iters = 1000

lambda_values = [0, 0.1, 1, 10, 100]
theta_values = {}

for lam in lambda_values:
    theta_trained, _ = gradientDescent(X, y, theta.copy(), alpha, best_num_iters, lam)
    theta_values[lam] = theta_trained
```

```python
plt.figure(figsize=(10, 6))
lambda_values = [0, 0.1, 1, 10, 100]
theta_values = {}

for lam in lambda_values:
    theta_trained, _ = gradientDescent(Xb, y, theta0.copy(), alpha, best_iter, lam)  # Xb burada!
    theta_values[lam] = theta_trained
for lam, theta_trained in theta_values.items():
    x_values = np.array([Xb[:, 1].min(), Xb[:, 1].max()])
    if len(theta_trained) >= 3 and not np.isclose(theta_trained[2], 0.0):
        y_values = -(theta_trained[0] + theta_trained[1] * x_values) / theta_trained[2]
        plt.plot(x_values, y_values, label=f'λ = {lam}')

plt.scatter(Xb[y == 1, 1], Xb[y == 1, 2], label="Outcome = 1", marker='o', color='b')
plt.scatter(Xb[y == 0, 1], Xb[y == 0, 2], label="Outcome = 0", marker='x', color='r')

plt.xlabel("Normalized Glucose")
plt.ylabel("Normalized BMI")
plt.title("Decision Boundaries for Different Lambda Values")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure()
for key, values in cost_histories.items():
    plt.plot(values, label=f"{key} iterations")
plt.xlabel("Iteration")
plt.ylabel("Cost")
plt.title("Cost vs Iteration (Unregularized)")
plt.legend()
plt.show()
```
